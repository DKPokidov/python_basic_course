## Конкатенация в Python

Конкатенация — это слово для операции склеивания или соединения чего-либо. Соединение требуется, когда вы хотите, например, распечатать разные типы данных при помощи одного print().

### 1. Конкатенация строк с помощью оператора +

```python
name = "Кот"
greeting = "Мяу, меня зовут " + name + "!"
print(greeting)
```

### 2. Конкатенация строк с помощью запятой

```python
name = "Кот"
print("Мяу, меня зовут", name, "!")
```

Важный момент: числа нельзя просто так склеить со строками. \
Сначала их нужно преобразовать в строку с помощью функции str().

Вот пример с ошибкой?

```python
age = 3
message = "Мне " + age + " года"  # Вызовет ошибку TypeError
print(message)
```

### 3. Конкатенация строк с помощью f-строк

Более современный способ. \
Они намного удобнее, особенно когда нужно склеить много разных данных.\
Переменные подставляются прямо внутрь строки в фигурных скобках {}.

```python
name = "Аристарх"
action = "программировать"
hours = 1.5

message = f"Студент {name} любит {action} ровно {hours} часа."
print(message)
```

### И зачем это все?

```python
name = input('Введите имя студента')
action = input('Что он любит делать?')
hours = float(input('Сколько часов подряд?'))

message = f"Студент {name} любит {action} ровно {hours} часа."
print(message)
```

Задача № 4*. Сочинитель сказок

Вы очень любите своего младшего брата (ну или хотя бы представьте, что он у вас есть или что вы его любите). \
Родители поручили вам читать ему сказки на ночь. Напишите программу, которая создаёт забавную историю с пропусками, заполняемыми словами пользователя. Программа должна последовательно выводить фрагменты текста и запрашивать у пользователя ввод слов для заполнения пропусков.

Вот пример работы такой программы: 

```python
# Приветствие
print("Привет! Давай создадим самую безумную историю про хомяка!\n")

# Первый фрагмент текста
name = input("1. Введите имя хомяка: ")
print(f"\nЖил-был хомяк по имени {name}")

# Второй фрагмент
food = input("\n2. Что он любил есть? ")
profession = input("3. Кем он мечтал стать? ")
print(f"\n{name} очень любил есть {food} и мечтал стать {profession}")

# Третий фрагмент
skill = input("\n4. Чему он научился? ")
place = input("5. Куда он отправился? ")
print(f"\nОднажды {name} научился {skill} и отправился в {place}")

# Четвертый фрагмент
character = input("\n6. Кто его там встретил? ")
quote = input(f"7. Что {character} сказал? ")
print(f"\nВ {place} его встретил {character}, который сказал: \"{quote}\"")

# Заключительный фрагмент
sentence = input("\n6. Что хомяк ответил ")
adjective = input("\n8. Каким он стал? ")
print(f'\n"{sentence}" — и с тех пор {name} стал самым {adjective} хомяком в мире!')
```
