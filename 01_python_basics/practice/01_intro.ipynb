### Арифметические действия

```python
a = 5
b = 4
print('a = ', a, '\n' 'b = ', b)
print('*************************')

a = a + b
print('a = ', a, '\n' 'b = ', b)
print('*************************')

b += a # b = b + a
print('a = ', a, '\n' 'b = ', b)
```

### Как работают сравнения?

```python
a = 4
b = 5.5

print(a < b)
print(a != b)
print(a >= b)
```
