# Визуализация данных в Python: Matplotlib

**Matplotlib** — популярная Python-библиотека для визуализации данных. Она используется для создания любых видов графиков: линейных, круговых диаграмм, столбчатых гистограмм и других — в зависимости от задач.

Matplotlib используют для визуализации данных любой сложности. Библиотека позволяет строить разные варианты графиков: линейные, трёхмерные, диаграммы рассеяния и другие, а также комбинировать их.

Подробно про Matplotlib можно узнать из [официальной документации](https://matplotlib.org/stable/users/index.html).

### Импорт библиотеки

Стандартный псевдоним для Matplotlib — `plt`.

## Строим первый график в Matplotlib

Подробно про настройку параметров функции `plt.plot()` можно прочесть в [официальной документации](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html).

## Диаграмма рассеяния (scatter plot)

Диаграмма рассеяния используется для оценки взаимосвязи двух переменных, значения которых откладываются по разным осям.

### Основные цвета в Matplotlib

| Символ | Цвет |
|--------|------|
| 'b' | синий (blue) |
| 'g' | зелёный (green) |
| 'r' | красный (red) |
| 'c' | голубой (cyan) |
| 'm' | пурпурный (magenta) |
| 'y' | жёлтый (yellow) |
| 'k' | чёрный (black) |
| 'w' | белый (white) |

Больше о цветовых форматах в [официальной документации](https://matplotlib.org/stable/users/explain/colors/colors.html).

## Столбчатая диаграмма (bar chart)

Такой тип визуализации позволяет удобно сравнивать значения отдельных переменных. В столбчатой диаграмме длина столбцов пропорциональна показателям, которые они отображают. Как правило, одна из осей соответствует одной категории, а вторая — её дискретному значению.

## Комбинирование разных типов графиков

На одном холсте можно отображать несколько типов графиков одновременно.

## Круговая диаграмма (pie chart)

Круговую диаграмму используют для отображения состава групп. Например, мы можем наглядно показать, какие марки автомобилей преобладают на дорогах города.

## Сложные визуализации: несколько подграфиков

Функция `plt.subplots()` позволяет создавать несколько графиков в одном окне.

## Столбчатая диаграмма с накоплением (stacked bar chart)

Позволяет оценить динамику соотношения значений одной переменной.
