## 2. Основные типы графиков Seaborn

### 2.1. Диаграмма рассеяния (scatterplot)

**Что это:** Точки на плоскости, каждая точка — одно наблюдение.

**Когда использовать:** Чтобы увидеть, есть ли связь между двумя числами. Например, зависит ли сумма чаевых от суммы счета? Чем больше счет — тем больше чаевые? Или нет?

**Где применяется:** В бизнесе, маркетинге, экономике — везде, где нужно понять связь между двумя показателями.

Документация: [sns.scatterplot](https://seaborn.pydata.org/generated/seaborn.scatterplot.html)

**Что мы видим:** В целом, чем выше счет, тем больше чаевые. Но есть и исключения — большие счета с маленькими чаевыми.

### 2.2. Ящик с усами (boxplot)

**Что это:** График, который показывает, как распределены данные.

**Как читать:**
- Линия внутри ящика — медиана (среднее значение, 50% данных слева, 50% справа)
- Нижняя граница ящика — 25-й перцентиль (25% данных ниже)
- Верхняя граница ящика — 75-й перцентиль (75% данных ниже)
- Усы (линии вверх и вниз) — разброс данных
- Точки за усами — выбросы (необычно большие или маленькие значения)

**Когда использовать:** Чтобы сравнить распределение одной величины в разных группах. Например, отличаются ли чаевые в понедельник и в субботу?

**Где применяется:** Медицина (сравнение показателей), контроль качества, анализ продаж.

Документация: [sns.boxplot](https://seaborn.pydata.org/generated/seaborn.boxplot.html)

**Что мы видим:** В выходные (суббота, воскресенье) чаевые в среднем выше, чем в будни. У женщин разброс чаевых меньше, чем у мужчин.

### 2.3. Гистограмма (histplot)

**Что это:** График, который показывает, как часто встречаются разные значения.

**Как читать:** Высота столбца показывает, сколько наблюдений попало в этот интервал.

**Когда использовать:** Чтобы понять, как распределены данные — есть ли много маленьких значений или много больших, симметрично ли распределение.

**Где применяется:** Анализ доходов, возраста клиентов, времени отклика сервера.

**Дополнительно:** Плавная линия (KDE — ядерная оценка плотности) — это сглаженная версия гистограммы, помогает лучше увидеть форму распределения.

Документация: [sns.histplot](https://seaborn.pydata.org/generated/seaborn.histplot.html)

**Что мы видим:** Большинство счетов — от 10 до 25 долларов. Ужины в среднем дороже обедов. Вечером чаще встречаются большие счета (30+ долларов).

### 2.4. Тепловая карта корреляции (heatmap)

**Что это:** Таблица, где цвет показывает силу связи между числами.

**Что такое корреляция:**
- Корреляция от -1 до 1 показывает, связаны ли две величины
- 1 — идеальная прямая связь (одно растёт — другое растёт)
- -1 — идеальная обратная связь (одно растёт — другое падает)
- 0 — связи нет

**Когда использовать:** Чтобы быстро найти, какие признаки влияют друг на друга.

**Где применяется:** Маркетинг (связаны ли расходы на рекламу и продажи), финансы, медицина.
