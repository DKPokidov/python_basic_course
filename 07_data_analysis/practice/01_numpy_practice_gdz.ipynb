Представьте, что у вас есть данные о температуре в городе за месяц (30 дней). Выполните следующие действия:

1. Создайте массив из 30 случайных чисел в диапазоне 15–30 °C (имитация температур).

2. Найдите среднюю температуру за месяц.

3. Найдите максимальную и минимальную температуру.

4. Определите, сколько дней температура была выше средней.

5. Замените все значения ниже 20 °C на 20 (предположим, это порог комфорта).

6. Выведите все результаты с пояснениями.

```python
# 1. Создаём массив температур
temperatures = np.random.uniform(15, 30, 30)
temperatures = np.round(temperatures, 1)  # Округляем до 1 знака после запятой

# 2. Средняя температура
avg_temp = np.mean(temperatures)

# 3. Макс. и мин. температура
max_temp = np.max(temperatures)
min_temp = np.min(temperatures)

# 4. Дни выше средней температуры
days_above_avg = np.sum(temperatures > avg_temp)

# 5. Замена температур ниже 20 °C
temperatures[temperatures < 20] = 20

# 6. Вывод результатов
print("Температуры за месяц (°C):", temperatures)
print(f"Средняя температура: {avg_temp:.1f} °C")
print(f"Максимальная температура: {max_temp:.1f} °C")
print(f"Минимальная температура: {min_temp:.1f} °C")
print(f"Дней с температурой выше средней: {days_above_avg}")
print("Температуры после замены (<20 °C → 20 °C):", temperatures)
```

#### Задача 5. Анализ успеваемости студентов
Представьте, что вы анализируете результаты экзаменационной сессии для группы из 25 студентов по трём предметам: математике, физике и программированию. Выполните следующие действия:

1. Создайте двумерный массив размером 25×3, заполненный случайными оценками от 2 до 5 (включительно). Каждая строка — это оценки одного студента, столбцы соответствуют предметам: математика, физика, программирование.

2. Рассчитайте средний балл каждого студента (по трём предметам) и добавьте эти значения в виде четвёртого столбца массива.

3. Найдите:

    * средний балл по каждому предмету (среднее по столбцам);

    * максимальную и минимальную оценку по каждому предмету;

    * количество студентов, получивших «5» по математике;

    * количество студентов, у которых средний балл ≥ 4.

4. Выделите «отличников» — студентов со средним баллом 4.5 и выше. Создайте отдельный массив с их оценками и средними баллами.

5. Для всего массива оценок (без учёта средних баллов) определите:

    * процент оценок «5»;

    * процент оценок «3» и «2» (неудовлетворительные оценки).

6. Выведите все результаты с понятными пояснениями.

```python
# 1. Создаём массив оценок (25 студентов × 3 предмета)
np.random.seed(42)  # для воспроизводимости результатов
grades = np.random.randint(2, 6, size=(25, 3))  # оценки от 2 до 5
subjects = ['Математика', 'Физика', 'Программирование']

print("Оценки студентов (25×3):")
for i, student_grades in enumerate(grades):
    print(f"Студент {i+1}: {student_grades}")

# 2. Рассчитываем средний балл для каждого студента
average_scores = np.mean(grades, axis=1, keepdims=True)  # среднее по строкам
full_data = np.hstack([grades, average_scores])  # добавляем столбец средних баллов

print(f"\nСредний балл каждого студента: {average_scores.flatten()}")

# 3. Анализируем данные по предметам
avg_per_subject = np.mean(grades, axis=0)  # среднее по столбцам
max_per_subject = np.max(grades, axis=0)
min_per_subject = np.min(grades, axis=0)

count_fives_math = np.sum(grades[:, 0] == 5)  # студенты с «5» по математике
count_high_avg = np.sum(average_scores >= 4)  # студенты со средним ≥ 4

print("\nАнализ по предметам:")
for i, subject in enumerate(subjects):
    print(f"{subject}:")
    print(f"  Средний балл: {avg_per_subject[i]:.2f}")
    print(f"  Максимальная оценка: {max_per_subject[i]}")
    print(f"  Минимальная оценка: {min_per_subject[i]}")

print(f"\nКоличество студентов с «5» по математике: {count_fives_math}")
print(f"Количество студентов со средним баллом ≥ 4: {count_high_avg}")

# 4. Выделяем «отличников» (средний балл ≥ 4.5)
excellent_students = full_data[full_data[:, 3] >= 4.5]
print(f"\nКоличество отличников (средний балл ≥ 4.5): {len(excellent_students)}")
if len(excellent_students) > 0:
    print("Оценки отличников (с средними баллами):")
    for i, row in enumerate(excellent_students):
        print(f"Отличник {i+1}: Математика={row[0]}, Физика={row[1]}, Программирование={row[2]}, Средний={row[3]:.2f}")

# 5. Процентное соотношение оценок
total_grades = grades.size  # общее количество оценок
percent_fives = np.sum(grades == 5) / total_grades * 100
percent_unsatisfactory = np.sum((grades == 2) | (grades == 3)) / total_grades * 100

print(f"\nПроцент оценок «5»: {percent_fives:.1f}%")
print(f"Процент неудовлетворительных оценок («2» и «3»): {percent_unsatisfactory:.1f}%")
```
