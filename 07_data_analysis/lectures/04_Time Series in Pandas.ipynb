Фильтрация по периодам. Проведем подготовительные процедуры

#### Ресемплирование данных

**Ресемплинг** (resampling, передискретизация) — это изменение частоты (периодичности) временных рядов: преобразование данных из одной временной шкалы в другую.

Метод *resample()* — изменение частоты временных рядов.

Частоты (freq):

* 'd' — день;
* 'h' — час;
* 'm' — месяц;
* 'q' — квартал;
* 'y' — год;

Когда использовать ресемплинг?  

* Анализ трендов: даунсемплинг помогает увидеть долгосрочные изменения (например, месячные продажи вместо ежедневных).
* Прогнозирование: модели часто требуют регулярной частоты (ежедневные данные вместо хаотичных записей).
* Сравнение: привести разные временные ряды к одной частоте для анализа.
* Визуализация: сгладить шум для понятных графиков.

##### Понижение частоты, Downsampling (например, с минут/часов на день)

**Что:** объединение данных с высокой частоты в более крупные периоды.  
**Зачем**: уменьшить объём данных, выделить тренды, сгладить шум.

Примеры:

* с минут на часы;

* с часов на дни;

* с дней на недели/месяцы.

```python
import matplotlib.pyplot as plt
import numpy as np

# Создаём синтетические данные: часовые данные за 3 дня
np.random.seed(42)  # для воспроизводимости
dates_hourly = pd.date_range('2024-01-01', '2024-01-03 23:00', freq='h')
hourly_values = 50 + 10 * np.sin(np.linspace(0, 4*np.pi, len(dates_hourly))) + np.random.randn(len(dates_hourly)) * 5
ts_plot = pd.Series(hourly_values, index=dates_hourly)

# Выполняем ресемплинг: часовые → дневные суммы
daily_sum = ts_plot.resample('d').sum()


# Визуализация
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 10), sharex=True)

# Верхний график: исходные часовые данные
ax1.plot(ts_plot.index, ts_plot.values, 'b-', alpha=0.6, linewidth=1, label='Часовые данные')
ax1.set_title('Исходные данные (часовые)', fontsize=14, fontweight='bold')
ax1.set_ylabel('Значение', fontsize=12)
ax1.grid(True, alpha=0.3)
ax1.legend()

# Нижний график: результат ресемплинга (дневные суммы)
ax2.bar(daily_sum.index, daily_sum.values, color='orange', alpha=0.7, width=0.8, label='Дневные суммы (resample)')
ax2.set_title('Результат ресемплинга (часовые → дневные, сумма за день)', fontsize=14, fontweight='bold')
ax2.set_xlabel('Дата', fontsize=12)
ax2.set_ylabel('Сумма за день', fontsize=12)
ax2.grid(True, alpha=0.3)
ax2.legend()

# Настройка общего внешнего вида
plt.tight_layout()
plt.show()
```

Пример 2. Повышение частоты, Upsampling (с дней на часы):

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np


# Создаём синтетические данные: ежемесячные значения за 3 месяца
np.random.seed(42)
dates_monthly = pd.date_range('2024-01-01', '2024-03-01', freq='MS')  # MS — начало месяца
monthly_values = [100, 150, 120]
ts_monthly = pd.Series(monthly_values, index=dates_monthly)

# Выполняем ресемплинг: ежемесячные → ежедневные (апсемплинг)
daily_ffill = ts_monthly.resample('D').ffill()      # прямое заполнение
daily_bfill = ts_monthly.resample('D').bfill()      # обратное заполнение
daily_interpolate = ts_monthly.resample('D').interpolate(method='linear')  # линейная интерполяция

# Визуализация
fig, axes = plt.subplots(2, 2, figsize=(16, 10))
fig.suptitle('Визуализация апсемплинга (повышение частоты: месячные → ежедневные данные)', fontsize=16, fontweight='bold')

# 1. Исходные месячные данные
axes[0, 0].bar(ts_monthly.index, ts_monthly.values, color='blue', alpha=0.7, width=25, label='Месячные данные')
axes[0, 0].set_title('Исходные данные (месячные)')
axes[0, 0].set_ylabel('Значение')
axes[0, 0].grid(True, alpha=0.3)
axes[0, 0].legend()

# 2. Прямое заполнение (ffill)
axes[0, 1].plot(daily_ffill.index, daily_ffill.values, 'orange', marker='o', markersize=3, label='FFill (прямое заполнение)')
axes[0, 1].set_title('Метод ffill(): повторяет последнее известное значение')
axes[0, 1].set_ylabel('Значение')
axes[0, 1].grid(True, alpha=0.3)
axes[0, 1].legend()

# 3. Обратное заполнение (bfill)
axes[1, 0].plot(daily_bfill.index, daily_bfill.values, 'green', marker='s', markersize=3, label='BFill (обратное заполнение)')
axes[1, 0].set_title('Метод bfill(): берёт следующее известное значение')
axes[1, 0].set_xlabel('Дата')
axes[1, 0].set_ylabel('Значение')
axes[1, 0].grid(True, alpha=0.3)
axes[1, 0].legend()

# 4. Линейная интерполяция
axes[1, 1].plot(daily_interpolate.index, daily_interpolate.values, 'purple', marker='^', markersize=3, label='Interpolate (линейная)')
axes[1, 1].set_title('Метод interpolate(): плавный переход между точками')
axes[1, 1].set_xlabel('Дата')
axes[1, 1].set_ylabel('Значение')
axes[1, 1].grid(True, alpha=0.3)
axes[1, 1].legend()

plt.tight_layout()
plt.show()
```

```python
hourly_filled = ts.resample('h').ffill()  # заполнение вперёд
hourly_bfilled = ts.resample('h').bfill()  # заполнение назад
hourly_filled
```

**Notes!**

При повышении частоты количество строк увеличивается (в 24 раза для 'H' относительно 'D'), что может вызвать проблемы с памятью для больших наборов данных.

#### Скользящие окна и анализ трендов
**Скользящее окно** (rolling window) — метод анализа данных, при котором функция последовательно применяется к последовательным подмножествам данных фиксированного размера.

Принцип работы:

* Выбирается размер окна (например, 3 дня).

* Функция (среднее, сумма и т. д.) вычисляется для первых 3 дней.

* Окно сдвигается на один период вперёд.

* Процесс повторяется до конца данных.

В Pandas для работы со скользящими окнами используется метод rolling():

```python
# Создаём данные с шумом
np.random.seed(42)
dates = pd.date_range('2024-01-01', periods=100, freq='D')
values = 50 + 10 * np.sin(np.linspace(0, 4*np.pi, 100)) + np.random.randn(100) * 3 # синусоида имитирует сезонность, а рандом добавляет шума
ts = pd.Series(values, index=dates)
ts
```

```python
# Скользящее среднее за 7 дней
rolling_mean = ts.rolling(window=7).mean()
rolling_mean.head(17)
```

```python
# Визуализация
plt.figure(figsize=(12, 6))
plt.plot(ts.index, ts.values, 'b-', label='Исходные данные')
plt.plot(rolling_mean.index, rolling_mean.values, 'r-', linewidth=2, label='Скользящее среднее (7 дней)')
plt.title('Скользящее среднее для сглаживания шума')
plt.xlabel('Дата')
plt.ylabel('Значение')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

#### Сдвиги и лаги

**Лаг** (от англ. lag — задержка) — это сдвиг временного ряда на определённое количество периодов назад.
Сдвиг — общее понятие, включающее:

* лаг (сдвиг назад, lagging);
* лид (сдвиг вперёд, leading).

Лаги позволяют:
* анализировать автокорреляцию (связь текущего значения с прошлыми);
* создавать признаки для моделей прогнозирования (использовать прошлые значения как входные данные);
* выявлять сезонность и циклы (сравнивать текущее значение с аналогичным периодом в прошлом);
* строить модели ARIMA и другие модели временных рядов;
* обнаруживать запаздывающие эффекты (например, влияние рекламы на продажи через несколько дней).

В Pandas для создания лагов используется метод *shift()*:

```python
ts.shift(periods=4)
```

Параметры:

* periods — количество периодов сдвига (положительное — назад, отрицательное — вперёд);
* fill_value — значение для заполнения пропусков (по умолчанию NaN).

```python
# Создаём тестовый временной ряд
dates = pd.date_range('2024-01-01', periods=10, freq='D')
values = [10, 15, 12, 18, 20, 16, 22, 25, 23, 28]
ts = pd.Series(values, index=dates)
print("Исходный ряд:")
ts
```

```python
# Лаг 1 (сдвиг на 1 период назад)
lag_1 = ts.shift(periods=1)
print("Лаг 1:")
print(lag_1)
print()
```

```python
# Лаг 3 (сдвиг на 3 периода назад)
lag_3 = ts.shift(periods=3)
print("Лаг 3:")
print(lag_3)
print()
```

```python
# Лид 1 (сдвиг на 1 период вперёд)
lead_1 = ts.shift(periods=-1)
print("Лид 1:")
print(lead_1)
```

1. Расчёт изменений (разностей)

```python
# Изменение за 1 период
change_1d = ts - ts.shift(1)

# Процентное изменение
pct_change = (ts / ts.shift(1) - 1) * 100

print("Изменения за 1 день:")
print(change_1d)
print("\nПроцентные изменения:")
print(pct_change)
```

2. Автокорреляция

**Автокорреляция** (англ. autocorrelation) — статистическая взаимосвязь между значениями одного и того же временного ряда, взятыми со сдвигом (лагом) по времени.

Проще говоря, автокорреляция показывает, насколько текущее значение временного ряда связано с его прошлыми значениями.

```python
lag = 6
autocorr = ts.corr(ts.shift(lag))
print(autocorr)
```

```python
# Автокорреляция с лагами 1–7
plt.figure(figsize=(10, 6))
for lag in range(1, 8):
    autocorr = ts.corr(ts.shift(lag))
    plt.bar(x=lag, height=autocorr)

plt.title('Автокорреляция по лагам (1–7 дней)')
plt.xlabel('Лаг (дни)')
plt.ylabel('Коэффициент корреляции')
plt.grid(True, alpha=0.3)
plt.show()
```

3. Выявление сезонности

```python
# Для недельной сезонности — лаг 7
weekly_seasonality = ts.corr(ts.shift(7))
print(f"Корреляция с недельной задержкой: {weekly_seasonality:.3f}")
```
