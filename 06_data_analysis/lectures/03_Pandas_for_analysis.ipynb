#### Поиск и обработка пропусков

**Что такое NaN?**

    NaN = Not a Number (не число)

Это специальное значение, которое обозначает отсутствие данных, пропуск или некорректное числовое значение. В Pandas и NumPy это стандартный способ представить, что ячейка пуста.

```python
import numpy as np

df = pd.DataFrame({'a': [1, np.nan, 3, None]})
print(df)
```

**Note!**

В Pandas есть несколько разных сущностей, которые обозначают пропуски. Все они работают по-разному.

1. np.nan

```python
value = np.nan
print(value) 
print(type(value)) 
```

Особенности:

* Это объект типа float

* *np.nan != np.nan* (одно пропущенное значение не равно другому!)

* При операциях "заражает" результат: 
    
    *1 + np.nan → np.nan*

```python
1 + np.nan
```

2. None (родной Python)

```python
value = None
print(type(value))
```

Особенности:

* Родной для Python, не из NumPy

* В Pandas автоматически преобразуется в np.nan в числовых колонках

* В объектных колонках (строки, смешанные типы) может остаться как None

```python
# None + 1
```

##### np.nan vs None — разница в двух словах

Методы для работы с пропусками:

* *isna() / isnull()* — поиск пропусков (возвращает булев массив);

* *notna() / notnull()* — поиск заполненных значений;

* *dropna()* — удаление строк/столбцов с пропусками;

* *fillna()* — заполнение пропусков.

```python
# Создаём DataFrame с пропусками
df = pd.DataFrame({
    'A': [1, 2, np.nan, 4, 5],
    'B': [np.nan, 2, 3, np.nan, 5],
    'C': ['x', 'y', np.nan, 'w', 'z']
})
```

```python
# Поиск пропусков
df.isna() #.sum()
```

```python
# Удаление строк с пропусками
df_cleaned = df.dropna()
print("\nПосле dropna():")
print(df_cleaned)
```

```python
# Заполнение пропусков разными способами
df_filled = df.copy()
df_filled['A'] = df_filled['A'].fillna(df_filled['A'].mean())  # средним
df_filled['B'] = df_filled['B'].fillna(df_filled['B'].median())  # медианой
df_filled['C'] = df_filled['C'].fillna('Неизвестно')  # константой
print("\nПосле fillna():")
print(df_filled)
```

*groupby()* группирует данные по значениям одного или нескольких столбцов. После группировки можно применять агрегатные функции.

##### Базовые агрегатные функции
После группировки можно применить стандартные агрегатные функции:

* *sum()* — сумма значений в группе;

* *mean()* — среднее значение;

* *count()* — количество элементов в группе;

* *max()* — максимальное значение;

* *min()* — минимальное значение.

```python
# Количество школ в каждой категории размера
schools.groupby('school_size').count()
```

```python
# Количество школ по размерам:
schools.groupby('school_size').size()
```

```python
schools.groupby('school_size')['capacity'].mean()
```

```python
# Максимальная площадь здания в каждой категории
schools.groupby('school_size')['building_area'].max()
```

```python
# Минимальная вместимость в каждой категории specific_capacity
schools.groupby('specific_capacity')['capacity'].min()
```

##### Множественная агрегация: *agg()* с разными функциями
Метод *agg()* позволяет применить несколько функций одновременно к одной или разным колонкам.

```python
# Разные функции для разных столбцов
result1 = schools.groupby('school_size').agg({
    'building_area': 'mean',
    'capacity': 'sum'
})
print("Множественная агрегация (средняя площадь и сумма вместимости):")
result1
```

```python
# Несколько функций для одного столбца
result2 = schools.groupby('specific_capacity')['building_area'].agg(['sum', 'mean', 'count'])
print("\nНесколько функций для площади зданий по specific_capacity:")
result2
```

```python
# Комбинированная агрегация с пользовательскими функциями
result3 = schools.groupby('school_size').agg({
    'building_area': ['sum', 'mean'],
    'capacity': ['count', 'min', 'max']
})

result3
```

##### Группировка по нескольким столбцам

```python
# Группировка по размеру школы и специфической вместимости
multi_group = schools.groupby(['school_size', 'specific_capacity'])
result4 = multi_group.agg({
    'building_area': 'sum',
    'capacity': 'mean'
})# Агрегация по двум уровням

result4
```

#### Объединение и слияние данных 

```python
# DataFrame 1: основные данные о школах
schools_main = pd.DataFrame({
    'object_name': ['Школа №1', 'Школа №2', 'Школа №3', 'Школа №4'],
    'building_area': [1000.0, 1200.5, 800.0, 1500.2],
    'capacity': [500, 600, 400, 700],
    'school_size': ['Большая', 'Средняя', 'Маленькая', 'Большая']
}, index=[0, 1, 2, 3])

# DataFrame 2: дополнительные данные о школах (продолжение)
schools_additional = pd.DataFrame({
    'object_name': ['Школа №5', 'Школа №6', 'Школа №7'],
    'building_area': [900.8, 1100.3, 1300.1],
    'capacity': [300, 550, 650],
    'school_size': ['Маленькая', 'Средняя', 'Средняя']
}, index=[4, 5, 6])

# DataFrame 3: данные по финансированию
schools_funding = pd.DataFrame({
    'object_name': ['Школа №1', 'Школа №2', 'Школа №5', 'Школа №8'],
    'budget_2023': [5.0, 6.2, 4.8, 7.1],  # млн руб.
    'funding_type': ['Государственное', 'Государственное', 'Муниципальное', 'Государственное']
}, index=[0, 1, 4, 7])

print("Основные данные о школах:")
print(schools_main)
print("\nДополнительные данные о школах:")
print(schools_additional)
print("\nДанные по финансированию:")
print(schools_funding)
```

##### Конкатенация: pd.concat() (вертикальная и горизонтальная)
Синтаксис:

    pd.concat([df1, df2, ...], axis=0/1, ignore_index=False)
    
* axis=0 — вертикальная конкатенация (добавление строк);

* axis=1 — горизонтальная конкатенация (добавление столбцов);

* ignore_index=True — сброс индекса.

```python
# Вертикальная конкатенация: объединение двух DataFrame с данными о школах
schools_all = pd.concat([schools_main, schools_additional])
print("Вертикальная конкатенация (все школы):")
schools_all
```

```python
# Горизонтальная конкатенация: объединяем данные о школах и финансировании по строкам (индексам)
schools_with_funding = pd.concat([schools_main, schools_funding], axis=1)
print("Горизонтальная конкатенация:")
schools_with_funding
```

**Note!**

Объединение происходит по ИНДЕКСАМ! Строки выравниваются по индексам.

##### Слияние таблиц: merge() (inner, outer, left, right join)
Слияние по значениям в столбцах

Синтаксис:

    pd.merge(left, right, on='key', how='inner')

1. *on* — столбец для слияния;

2. *how* — тип соединения:

  * *inner* — только общие строки; 

  * *outer* — все строки из обеих таблиц;

  * *left* — все строки из левой таблицы;

  * *right* — все строки из правой таблицы.

```python
# Inner join: только школы, которые есть в обеих таблицах
inner_merge = pd.merge(schools_main, schools_funding, on='object_name', how='inner')
print("Inner join (школы с финансированием):")
inner_merge
```

```python
# Left join: все школы из основной таблицы + финансирование, если есть
left_merge = pd.merge(schools_main, schools_funding, on='object_name', how='left')
print("\nLeft join (все основные школы + финансирование):")
left_merge
```

```python
# Right join: все школы из таблицы финансирования + данные, если есть
right_merge = pd.merge(schools_main, schools_funding, on='object_name', how='right')
print("\nRight join (все финансируемые школы + данные):")
right_merge
```

```python
# Outer join: все школы из обеих таблиц
outer_merge = pd.merge(schools_main, schools_funding, on='object_name', how='outer')
print("\nOuter join (все школы из обеих таблиц):")
outer_merge
```
