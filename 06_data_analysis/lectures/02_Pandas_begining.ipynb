### Series и DataFrame
Данные в Pandas представлены в двух видах: *Series*#* и *DataFrame*. Разберёмся с каждым из них.

#### Что такое Series?
*pd.Series* — это одномерный помеченный массив в библиотеке pandas для Python. Он может хранить данные любого типа (числа, строки, объекты Python и т. д.) и состоит из двух основных компонентов:

    * данных (значения);

    * индекса (метки для идентификации элементов).

Основные компоненты

1. Данные (*values*)

    * хранятся в виде массива *NumPy*;

    * могут быть любого типа: целые числа, числа с плавающей запятой, строки, даты, объекты;

    * после создания размер фиксирован.

2. Индекс (*index*)

    * последовательность меток для идентификации каждого элемента;

    * по умолчанию: целые числа от 0 до N−1;

    * можно задать произвольные метки (строки, даты и т. д.);

    * метки могут повторяться, но должны быть хешируемыми.

3. Тип данных (*dtype*)

    * автоматически определяется при создании, но можно задать явно;

    * основные типы: *int64, float64, object* (для строк и смешанных данных), *bool, datetime64*.

4. Имя (*name*)

    * необязательный атрибут — имя серии.

```python
# Создадим простой Series из списка:
s1 = pd.Series([1, -3, 5, 6], name='Первая')
s1
```

```python
# Из словаря (ключи → индекс, значения → данные)
s2 = pd.Series({'a': 10, 'b': 20, 'c': 30})
s2
```

```python
# Из массива NumPy
import numpy as np

arr = np.array([1.1, 2.2, 3.3])
s3 = pd.Series(arr)
s3
```

```python
# Из скалярного значения (повторяется для всех индексов)
s4 = pd.Series(5, index=['x', 'y', 'z'])
s4
```

#### Доступ к данным

```python
# По позиции (как в списке):
s1[0]  
s1[1:3]  
```

```python
# По метке индекса:
s2['a']
s2[['a', 'c']]
```

Методы *loc[ ]* и *iloc[ ]*:

    * iloc[]* — доступ по позициям: s1.iloc[0].

Ключевые свойства

    s.index — получить индекс;

    s.values — получить массив данных;

    s.dtype — тип данных;

    s.name — имя серии;

    len(s) — длина серии;

    s.shape — размерность (кортеж);

    s.size — количество элементов.

```python
print(f's1.index: {s1.index}')
print(f's1.values: {s1.values}')
print(f's1.dtype: {s1.dtype}')
print(f's1.name: {s1.name}')
print(f'len(s1): {len(s1)}')
print(f's1.shape: {s1.shape}')
print(f's1.size: {s1.size}')
```

#### Обработка пропущенных данных

1. Пропущенные значения обозначаются как *NaN* (Not a Number);

2. Автоматически исключаются из статистических операций (*sum( )*, *mean( )* и т. д.);

3. Методы для работы с *NaN*:

    * *isnull( )* / *notnull( )* — проверка на пропуски;

    * *dropna( )* — удалить пропуски;

    * *fillna(value)* — заполнить пропуски заданным значением.

```python
s_nan = pd.Series([1, np.nan, 3])
s_nan.isnull()
```

```python
# Почему ничего не изменилось?
s_nan
```

**Note!**

По умолчанию метод *fillna( )* в pandas не изменяет исходный объект, а возвращает новый — копию с заполненными значениями. Это поведение задаётся параметром *inplace=False* (значение по умолчанию).

```python
s_nan = pd.Series([1, np.nan, 3])
s_nan.isnull()
s_nan
```

```python
print(s_nan.dropna())
print(s_nan)
```

```python
s_nan.dropna(inplace=True)
s_nan
```

```python
s_nan.fillna(0)
```

#### Операции над Series
1. Арифметические (+, -, *, /, **)

    * выравнивают данные по меткам индекса;

    * если метки не совпадают, результат содержит NaN.

```python
a = pd.Series([1, 2], index=['a', 'b'])
b = pd.Series([3, 4], index=['b', 'c'])
a + b
```

```python
a.add([1, 2])
```

```python
d = pd.Series([1, -7, 3, 5])
d.abs()
```

2. Статистические (*sum( ), mean( ), std( ), min( ), max( )*)

    * игнорируют NaN по умолчанию.

```python
s7 = pd.Series([7, np.nan, 9, 2, 1])
print(f'Результат функции sum(): {s7.sum()}')
print(f'Результат функции mean(): {s7.mean()}')
print(f'Результат функции std(): {s7.std():.2f}')
print(f'Результат функции min(): {s7.min()}')
```

3. Применение функций (*apply(func)*)

    * применяет функцию к каждому элементу.

```python
temperatures = pd.Series(
    [20, 22, 19, 25],
    index=['Пн', 'Вт', 'Ср', 'Чт'],
    name='Температура (°C)'
)
temperatures
```

```python
# Температуры выше 20°C
hot_days = temperatures[temperatures > 20]
```

```python
# Перевод в Фаренгейты
fahrenheit = temperatures.apply(lambda c: c * 9/5 + 32)
```

```python
ser.sort_index()
```

5. Группировка и агрегация (*groupby( ), agg( )*)

```python
# группировка по значению
s8 = pd.Series([1, 5, 1, 4, 5], index=['b', 'a', 'c', 'b', 'a'])
s8.groupby(s8).sum()
```

```python
# группировка по индексу
grouped = s8.groupby(s8.index)
grouped.sum()
```

```python
# группировка по внешнему списку
categories = ['X', 'Y', 'X', 'Z', 'Y']
grouped = s8.groupby(categories)
grouped.sum()
```

```python
# группировка при помощи функции
s9 = pd.Series([x for x in range(10)])
grouped = s9.groupby(lambda x: 'четное' if x % 2 == 0 else 'нечетное')
grouped.count()
```

#### Что такое DataFrame?

**pd.DataFrame** — основной тип данных в Pandas, вокруг которого строится вся работа. Его можно представить в виде обычной таблицы с любым количеством столбцов и строк. Внутри ячеек такой «таблицы» могут быть данные самого разного типа: числовые, булевы, строковые и так далее.

DataFrame — двумерная помеченная структура данных с индексированными строками и столбцами (похожа на таблицу Excel). Состоит из:

* набора Series (каждый столбец — это Series);

* индекса строк (index);

* названий столбцов (columns).

1. Создание DataFrame из pd.Series

```python
s1 = pd.Series(['Москва', 'Санкт-Петербург', 'Новосибирск', 'Екатеринбург'], name='Город')
s2 = pd.Series([1147, 1703, 1893, 1723], name='Год основания')
df = pd.concat([s1, s2], axis=1)
df
```

Мультииндексы

**Мультииндекс (иерархический индекс)** — индекс с несколькими уровнями. Полезен для сложных данных (например, временные ряды по регионам и категориям).

```python
df.columns
```

```python
# Через set_index() с несколькими столбцами:
df_multi = df.set_index(['id', 'type'])
df_multi
```

```python
# Явное создание через pd.MultiIndex:
arrays = [['A', 'A', 'B', 'B'], ['x', 'y', 'x', 'y']]
multi_index = pd.MultiIndex.from_arrays(arrays, names=('level1', 'level2'))
df_custom = pd.DataFrame({'Значение': [1, 2, 3, 4]}, index=multi_index)
df_custom
```

Работа с мультииндексом

```python
# Доступ по уровням
df_custom.loc[('A', 'y'), 'Значение']
```

```python
# Выбор по первому уровню
df_custom.loc['B']
```

```python
# Свёртка по уровню
df_custom.groupby(level='level1').sum()
```

```python
# Свёртка по уровню
df_custom.groupby(level='level2').mean()
```
