#### Часть 4. Работа со сложными столбцами

Некоторые столбцы хранят словари в виде строк. Например, `'Id-общая видовая принадлежность'` содержит JSON-подобные строки. Нужно извлечь оттуда нужное значение `'value'`.

```python
# Посмотрим, что в столбце 'Id-общая видовая принадлежность'
okn_cleaned['Id-общая видовая принадлежность'].head(10)
```

#### Задание 3

Напишите функцию `extract_value()`, которая:
- Принимает значение из столбца.
- Если значение — NaN, возвращает None.
- Если значение — строка, похожая на словарь (содержит `{` и `}`), преобразует её в словарь с помощью `ast.literal_eval()` и возвращает значение по ключу `'value'`.
- Если строка — список словарей (начинается с `[`), возьмите первый словарь и извлеките `'value'`.
- В случае ошибки возвращает None.

Затем примените эту функцию к столбцу `'Id-общая видовая принадлежность'` и сохраните результат в новый столбец `'Видовая принадлежность'`.

```python
def extract_value(value):
    if pd.isna(value):
        return None
    
    try:
        value_str = str(value).strip()
        if value_str.startswith('[') and value_str.endswith(']'):
            dict_list = value_str[1:-1]
        else:
            dict_list = value_str
        
        if not dict_list:
            return None
        
        dictionary = ast.literal_eval(dict_list)
        
        if isinstance(dictionary, dict):
            return dictionary.get('value')
        elif isinstance(dictionary, tuple):
            return 'Памятник двойной принадлежности'
        else:
            return None
            
    except (ValueError, SyntaxError, TypeError, AttributeError) as e:
        print(f"Ошибка: {e} в значении: {value}")
        return None
```

```python
okn_cleaned['Видовая принадлежность'] = okn_cleaned['Id-общая видовая принадлежность'].map(extract_value)
okn_cleaned['Видовая принадлежность']
```

#### Задание 9

**Вопрос:** Найдите все объекты, в названии которых есть слово "Дорога жизни".
