## Введение в типы данных: строки, списки, кортежи, словари

Типы данных — одна из основных концепций в программировании. Именно они определяют характеристики переменных и указывают, какие операции можно выполнять с теми или иными данными.

В каждом из языков программирования есть свой набор заранее определенных типов данных, которые можно применять для хранения информации.

Типы данных нужны для того, чтобы:

* Обеспечить корректную работу ПО. Если вы задает тип данных для той или иной переменной, программа будет понимать, как с ними работать. Это дает возможность избежать ошибок, связанных с несоответствием типов данных. 
* Оптимизировать ресурсы. В памяти компьютера каждый из типов данных занимает определенное количество места. Если выбрать их правильно, то объем нужных ресурсов будет минимальным, а если использовать типы данных неправильно, то программа будет требовать максимальных ресурсов. 
* Писать понятный код. Типы данных указывают на то, какую информацию хранят и как ее можно применять. За счет их наличия программистам проще ориентироваться в написанном коде.
* Обеспечивать безопасность. За счет типов данных программы могут правильно применять ограничения к использованию информации. А это уменьшает уязвимость.

* Неизменяемые (немутабельные, immutable) типы данных: None, bool, int, float, complex, str, tuple.
* Изменяемые (мутабельные, muttable) типы данных: list, dict, set.

### Неизменяемые типы данных

```python
a = 100
b = a
print(f"a: id({id(a)})")
print(f"b: id({id(b)})")
```

Если мы попытаемся переменной a присвоить другое значение, то после операции += переменная будет ссылаться на другой объект в памяти.

```python
a = 100
b = a
print(f"a: id({id(a)})")
print(f"b: id({id(b)})")
a += 1
print(f"a: id({id(a)})")
```

#### Итак!
Неизменяемые типы данных в Python — это объекты, которые нельзя изменить после создания. Любая операция, предполагающая изменение, приводит к созданию нового объекта.

### Изменяемые типы данных

```python
list_a = [100]
list_b = list_a
print(f"list_a: id({id(list_a)})")
print(f"list_b: id({id(list_b)})")
list_a.append(101)
print(f"list_a: id({id(list_a)})")
print(f"list_b: id({id(list_b)})")
```

Суть здесь в том, что при попытке добавления элемента в изменяемый список не будет создан новый объект, а лишь произойдёт дополнение текущего.



#### Итак!
Изменяемые типы данных в Python — это объекты, содержимое которых можно исправлять после их создания. Эти изменения происходят без создания нового объекта, что означает, что все ссылки отражают его текущее состояние.

### Динамическая типизация данных

Под «динамической» типизацией подразумевается, что типы объектов определяются в процессе исполнения программы. Поэтому типы переменных указывать не обязательно, но это хороший тон.

Утиная типизация: если это выглядит как утка, плавает как утка и крякает как утка, то это, вероятно, утка. 



```python
a = input('Введите число: ')
print(a, ',', type(a))
```

```python
a = int(input('Введите число: '))
print(a, ',', type(a))
```

### Строки (Strings)
Строки — это последовательность символов, заключенная в кавычки.

```python
first_str = 'Университет '
second_str = 'ИТМО'
```

```python
#конкатенация строк
some_string = first_str + second_str
print(some_string)
```

```python
#длина строки
length = len(some_string)
print(length)
```

```python
#первый символ строки
first_symbol = some_string[0]
print(first_symbol)
```

```python
#срез строки с первого по пятый символ
itmo_slice = some_string[11:16]
print(itmo_slice) 
```

```python
#последний символ строки
last_symbol = some_string[-1]
print(last_symbol)
```

```python

#последние три символа строки
itmo_slice = some_string[-3:]
print(itmo_slice) #box
```

```python
st = 'Python'
print(st[0:6:2])
```

```python
#поиск подстроки
#если подстрока найдена, метод вернёт позицию её первого элемента, если нет — вернёт -1
ill_substr = some_string.find('ИТМ')
print(ill_substr)
```

```python
#поиск подстроки с заменой, заменим Университет на ИДУ
print(some_string)
idu_string = some_string.replace('Университет', 'ИДУ')
print(idu_string) 
```

```python
#разбить строку по разделителю
student_name = 'Ivan_Ivanovich_Petrov'
list_of_substr = student_name.split('_')
print(list_of_substr)
```

```python
# Удаляет пробелы в начале и конце строки, но может удалять другие указанные символы
student_name = '  Ivan Ivanovich Petrov  '
strip_student_name = student_name.strip()
print(strip_student_name)
```

```python
student_name = 'THOMAS ANDERSON'
student_name = student_name.lower() 
Student_name = student_name.capitalize()
STUDENT_NAME = student_name.upper()
print(student_name)
print(Student_name)
print(STUDENT_NAME)
```

Строки в Python являются неизменяемым типом данных (immutable), и это фундаментальное свойство объясняет, почему методы вроде lower(), upper(), strip() и другие возвращают новые строки, а не модифицируют существующую.

Больше о строках в документации: https://docs.python.org/3/library/string.html#

### Списки (Lists)
Список (list) — это упорядоченная коллекция объектов. \
Списки могут иметь сколько угодно уровней вложенности и хранить неограниченное количество объектов. Кроме того, в одном списке могут одновременно храниться объекты разных типов.

Над списками можно производить те же операции, что и над строками.

```python
first_list = ['cat', 123, True]
print(first_list)
```

#### Не называйте списки просто list
Если есть необходмость (или не хватает фантазии) назовите list_ или _list

```python
# Также в Python можно создавать вложенные списки
second_list = ['Москва', 2.12, [False, 4]]
print(second_list[2][0])
```

Важно помнить две вещи:

* у каждого элемента списка есть свой индекс;
* индексы начинаются с 0.

```python
third_list = [1, 'Нео', 3.14, True]

print(third_list[0]) # Обратится к 1
print(third_list[2]) # Обратится к 3
print(third_list[3]) # Обратится к 4
print(third_list[4]) # Выведет ошибку
```

```python
# здесь точно так же, как и со строками, работают срезы
print(third_list[1:3])
```

```python
# замена элемента
list_for_twist = [1, 'Черная кошка', 3]
list_for_twist[1] = 'Белый хомяк'
print(list_for_twist)
```

```python
# Объединение списков
a = [1, 2]
b = [3, 4]
с = a + b
print(с)
```

```python
# Разложение списка
print(third_list)
```

```python
number, name, pi, bools = third_list

print(number) # type()
print(name) # type()
print(pi)
print(bools)
```

```python
# Сравнение списков
a = [1, 2, 3]
b = [1, 2, 3]
print(a == b)
```

В некоторых языках равенство ещё проверяется и по тому, ссылаются ли переменные на один и тот же объект. Обычно это делается через оператор ===. В Python это можно сделать через оператор is, который проверяет, имеют ли две переменные один и тот же адрес в памяти:

```python
a = [1, 2, 3]
b = a
print(a is b)
```

##### Встроенные функции для списков Python

```python
# Возвращает длину списка
list_ = [5, 3, 1]
print(len(list_))
```

```python
# Возвращает отсортированный список
list_ = [8, 1, 3, 2]
print(sorted(list_))
```

```python
# Возвращают наименьший и наибольший элемент списка
list_ = [1, 9, -2, 3]
print(min(list_))
print(max(list_))
```

##### Методы списков Python

```python
# Добавляет новый элемент в конец списка
append_list = [1, 2, 3]
append_list.append(4)
print(append_list) 
```

```python
# Добавляет набор элементов в конец списка
extend_list = [1, 2, 3]
extend_list.extend([4, 5])
print(extend_list) 
```

```python
# Заметьте, что строка добавилась посимвольно
extend_list = ['cat', 'dog', 'bat']
extend_list.extend('mouse')
print(extend_list)
```

```python
# Добавляет новый элемент по индексу
insert_list = ['Тринити', 2, 3.14]
insert_list.insert(0, 'Нео')
print(insert_list)
```

```python
# Удаляет элемент из списка
insert_list.remove('Тринити')
print(insert_list)
```

```python
# Возвращает индекс элемента списка
print(third_list)
print(third_list.index(3.14))
```

```python
# Считает, сколько раз элемент повторяется в списке
count_list = [1, 1, 1, 2]
print(count_list.count(1))
```
