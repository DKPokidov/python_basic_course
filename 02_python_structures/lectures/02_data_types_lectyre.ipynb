## Кортежи
Кортежи (tuple) в Python хранят последовательности элементов, и этим они похожи на списки. Но между этими двумя структурами есть серьёзные различия. Сегодня мы разберёмся в них и научимся работать с кортежами.

putin_kortej.jpg

**Кортежи** (тип tuple) — это неизменяемый тип данных в Python, который используется для хранения упорядоченной последовательности элементов.

Свойства:
* **Неизменяемость**. После того как кортеж создан, в него нельзя добавлять элементы, а также изменять их или удалять. При попытке это сделать интерпретатор выдаст ошибку TypeError.

* **Упорядоченность**. Элементы кортежа располагаются в определённом порядке, который тоже неизменяем. К любому элементу можно обратиться по его индексу (порядковому номеру).

* Элементами кортежа **могут быть объекты разных типов** данных: числа, строки, списки, другие кортежи и другие. Элементы-коллекции могут иметь неограниченную глубину вложенности. Например, кортеж может включать в себя список, который будет содержать другой список, который вновь будет содержать список и так далее.

```python
my_tuple = (2.5, 'ИТМО', True)
print(my_tuple)
```

```python
# Чтобы обратиться ко вложенным элементам, используйте двойную индексацию:
nested_tuple = (1, 2, (3, 4), 5)


# Обращение к элементам кортежа верхнего уровня
print(nested_tuple[0])  # Вывод: 1
print(nested_tuple[3])  # Вывод: 5

print('='* 100)

# Обращение к элементам вложенного кортежа (3, 4)
print(nested_tuple[2])  # Вывод: (3, 4)
print(nested_tuple[2][0])  # Вывод: 3
print(nested_tuple[2][1])  # Вывод: 4
```

### Когда использовать кортеж
Применение кортежей связано с их свойствами — неизменяемостью и строгим порядком элементов. Разберём ситуации, когда это полезно.

**Требуется защитить элементы от изменений**

Если нужно защитить данные от случайного или намеренного изменения, то можно использовать кортежи. Например, этот тип данных подойдёт для сохранения информации о датах рождения клиентов.

Обратите внимание: неизменяемость кортежей не абсолютна. У неё есть исключение — если внутри кортежа находятся изменяемые элементы, например списки, словари или множества, то их значения можно изменить.

**Необходимо экономить память**

Кортежи занимают в памяти меньше места, чем списки. Это легко проверить:

```python
a = (10, 20, 30, 40, 50, 60) # Кортеж
b = [10, 20, 30, 40, 50, 60] # Список


print(a.__sizeof__()) # Размер кортежа в байтах
print(b.__sizeof__()) # Размер списка в байтах
```

### Упаковка и распаковка

Упаковка (packing) и распаковка (unpacking) — это операции, которые позволяют создавать кортеж из набора значений и извлекать значения из кортежа в переменные.

```python
# Упаковка кортежа
my_tuple = 1, 2, 3, 'hello'
# В переменной my_tuple теперь находится кортеж (1, 2, 3, 'hello')
```

```python
# Распаковка кортежа
a, b, c, d = my_tuple

# Теперь переменные a, b, c, d содержат значения из кортежа
print(a, b, c, d, sep=', ')
```

```python
# Обмен значениями между переменными
x = 5
y = 10

# Обмен значениями с использованием кортежа, упаковки и распаковки
x, y = y, x

# Теперь переменные x и y содержат новые значения
print('x =', x)
print('y =', y)
```

### Сравнение

Кортежи в Python сравниваются покомпонентно начиная с первого элемента до первого неравенства или до конца переменной. Если кортежи имеют одинаковую длину и все соответствующие элементы равны, то они считаются равными:

```python
# Пример 1: кортежи с числовыми значениями
tuple1 = (1, 2, 3)
tuple2 = (1, 2, 3)
print(tuple1 == tuple2)  # True, так как все элементы равны
```

```python
# Пример 2: кортежи с разными значениями
tuple3 = (4, 5, 6)
print(tuple1 == tuple3)  # False, так как первые элементы различаются
```

```python
# Пример 3: Сравнение кортежей разной длины
tuple4 = (1, 2, 3, 4)
print(tuple1 == tuple4)  # False, так как длина кортежей разная
```

Для сравнения используются стандартные операторы ==, !=, >, <, >=, <=.

При сравнении используются правила:

* когда сравниваются числовые элементы, большим считается тот кортеж, у которого первый отличающийся элемент больше;
* когда сравниваются строки, сравнение происходит по порядку в алфавите;
* элементы различных типов сравнивать нельзя — получим ошибку TypeError;
* если длины кортежей различны, то кортеж с меньшей длиной считается наименьшим.

```python
tuple1 = (1, 2, 3)
tuple2 = (1, 2, 4)

print(tuple1 < tuple2)  # True, так как tuple1 меньше tuple2 — (tuple1[3] < tuple2[3])
```

```python
tuple3 = ('apple', 'banana')
tuple4 = ('apple', 'orange')
print(tuple3 < tuple4)  # True, так как tuple3 «меньше» tuple4 по порядковому номеру первой отличающейся буквы
```

```python
tuple7 = (1, 2, 3)
tuple8 = (1, 2, 3, 4)


print(tuple7 < tuple8)  # True, так как tuple7 «меньше» tuple8 (кортеж меньшей длины меньше)
```

```python
tuple5 = (1, 2, 3)
tuple6 = ('apple', 'banana', 'orange')


print(tuple5 < tuple6)  # Ошибка TypeError: '<' not supported between instances of 'int' and 'str' — нельзя сравнивать числа и строки
```

### Как можно изменить элементы?



```python
# Трюк № 1. Преобразование кортежа в список
my_tuple = (1, 2, 3, 4, 5)
my_list = list(my_tuple)

# Изменение значения в списке
my_list[2] = 10

# Преобразование списка обратно в кортеж
updated_tuple = tuple(my_list)

print('Исходный кортеж:', my_tuple)
print('Обновлённый кортеж:', updated_tuple)
```

## Множества
**Множество (set)** — это изменяемый набор уникальных и неупорядоченных элементов.

Перечислим основные свойства множеств:

* **Уникальность**: каждый элемент множества неповторим. Если попробовать создать в наборе дубликат, Python не даст этого сделать.
* **Неупорядоченность**: у элементов множества нет порядкового номера.
* **Изменяемость**: можно добавлять во множество или удалять из него элементы.

```python
my_set = {1, 2.3, 'три', (3, 4, 4, 5)}
print(my_set)
```

Множества могут содержать только неизменяемые объекты:

* целые числа (integer);
* числа с плавающей точкой (float);
* строки (string);
* кортежи (tuple).

#### Создание множеств

```python
# с помощью фигурных скобок
my_set = {1, 2, 2, 3, 4, 4, 5}
print(my_set)
```

```python
# с помощью функции set()
my_set = set([1, 2, 3, 4, 4, 5, 5])
print(my_set)
```

```python
# из строк
my_string = "Hello, World!"
my_set = set(my_string)
print(my_set)
```

```python
# из кортежа
my_tuple = (1, 2, 2, 3, 4, 5, 5)
my_set = set(my_tuple)
print(my_set)
```

```python
# из диапазона
my_range = range(1, 6)
my_set = set(my_range)
print(my_set)
```

```python
# Из числа множество создать не удастся
my_num = 101598723
my_set = set(my_num)
```

```python
# предварительно надо будет превратить его в строку
my_num = 101598723
my_set = set(str(my_num))
print(my_set)
```

```python
# создание замороженного множества - добавить в него или удалить из него элементы не выйдет
my_set = frozenset([1, 2, 3, 4, 5])
print(my_set)
```

#### Как добавить элементы в множество?

```python
# добавляем в множество один объект
s = {1, 2, 3}
s.add(4) # используем функцию add()
print(s)
```

```python
# добавляем сразу несколько элементов
s = {1, 2, 3}
s.update([4, 5, 6]) # Аргументом может быть список, строка
print(s)
```

```python
# Множества в Python можно объединять друг с другом
set_1 = {1, 2, 3}
set_2 = {3, 4, 5}
set_3 = set_1.union(set_2) # используем функцию union()
print(set_3)
```

#### Удаление элементов множества

```python
# Метод remove() удаляет элемент из множества
s = {1, 2, 3, 4, 5}
s.remove(3)
print(s)
```

```python
# Если такого элемента в наборе не окажется, интерпретатор выдаст исключение
s.remove(8)
print(s)
```

```python
# Метод discard() работает безопаснее — если элемент не найден, исключения не выпадет
s = {1, 2, 3, 4, 5}
s.discard(8)
print(s)
```

```python
# Метод pop() удаляет случайный компонент множества
s = {1, 2, 3, 4, 5}
print(s.pop())
print(s)
```

```python
# метод clear() - полностью очищает множество
s = {1, 2, 3, 4, 5}
s.clear()
print(s)
```

#### Что запомнить

* Множество в Python — это изменяемая структура данных, которая содержит уникальные и неупорядоченные элементы. Множества полезны в случаях, когда нужно быстро проверить наличие элемента или удалить дубликаты из больших объёмов данных.

* Множества в Python могут содержать только неизменяемые типы данных — целые числа, строки и кортежи. Словари, списки и другие mutable-объекты не могут быть частью множества.

* Так как элементы множества неупорядоченны, вы не сможете получить доступ к ним по индексу — как, например, к элементам в списков и кортежей.

* Тип данных set поддерживает математические операции над множествами (например, объединение, пересечение, разность).

## Словари

Словарь в языках программирования — это что-то вроде телефонной книги, где под каждым номером скрывается какой-то человек. Только номера называют ключами, а людей, которым они принадлежат, — значениями. Вот как это можно представить в виде таблицы:

image.png

При этом ключи в словарях уникальны, а значения могут повторяться.

**Словарь — это неупорядоченная структура данных**.

Это значит, что все пары «ключ — значение» хранятся в произвольном порядке. Идея в том, что нам неважно, где находится элемент: в начале, в серединие или в конце. Важно то, что он лежит где-то внутри и мы можем при случае его достать.

```python
# Как создать словарь
books = {
  'Гарри Поттер и философский камень' : 'Джоан Роулинг',
  'Убить пересмешника' : 'Харпер Ли'
}
print(books)
```

```python
books = {
  'Гарри Поттер и философский камень',
  'Убить пересмешника'
}
print(type(books))
```

```python
# Как достать значение ключа из словаря
dict = {
  'Гарри Поттер и философский камень' : 'Джоан Роулинг',
  'Убить пересмешника' : 'Харпер Ли'
}

Harry_Potter_Author = dict['Гарри Поттер и философский камень']
print(Harry_Potter_Author)
```

```python
#  что будет, если не правильно указать ключ?
Harry_Potter_Author = dict['Гарри Поттер и бриллиантовый камень']

print(Harry_Potter_Author)
```

```python
# добавиляем новый элемент в словарь
dict['Преступление и наказание'] = 'Фёдор Достоевский'
print(dict)
```

```python
# удаляем элемент из словаря - используем команду del
del dict['Убить пересмешника']
print(dict)
```

```python
# изменяем значение ключа в словаре
dict['Гарри Поттер и философский камень'] = 'Михаил Лермонтов'
print(dict)
```

```python
# а что произойдет сейчас?
dict['Барри Воттер и философский камень'] = 'Михаил Лермонтов'

print(dict)
```

#### Методы для работы со словарями

```python
# Метод update() добавляет в словарь одну или сразу несколько пар «ключ — значение»
dict = {
  'Гарри Поттер и философский камень' : 'Джоан Роулинг',
  'Убить пересмешника' : 'Харпер Ли'
}
dict.update({
    'Преступление и наказание' : 'Фёдор Достоевский',
    'Чемодан' : 'Сергей Довлатов'
})

print(dict)
```

```python
# Метод get() возвращает значение из словаря по ключу
Harry_Potter_Author = dict.get('Гарри Поттер и философский камень')
print(Harry_Potter_Author)
```

```python
# а если такого ключа нет?
Harry_Potter_Author = dict.get('Гарри Поттер в путешествии по Золотому кольцу')
print(Harry_Potter_Author)
```

```python
# второй аргумент позволяет вывести любое значение, которое вы захотите
Harry_Potter_Author = dict.get('Гарри Поттер в путешествии по Золотому кольцу', 'Такого ключа нет')
print(Harry_Potter_Author)
```

```python
# Метод pop() удаляет элемент из словаря по ключу
Harry_Potter_Author = dict.pop('Гарри Поттер и философский камень')
print(Harry_Potter_Author)
print(dict)
```

```python
# Метод keys() возвращает все ключи из словаря
dict_keys = dict.keys()
print(dict_keys)
```

```python
# Метод values() возвращает все значения из словаря
dict_values = dict.values()
print(dict_values)
```

```python
# Метод items() возвращает все пары «ключ — значение»
dict_items = dict.items()
print(dict_items)
```
