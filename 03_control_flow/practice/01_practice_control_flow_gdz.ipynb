#### Задача 1. Проверка категории здания

Программа запрашивает тип здания (например, «школа», «торговый центр», «жилой дом») и определяет его категорию:

* «школа», «детский сад», «университет» → «Образовательное учреждение»;

* «торговый центр», «магазин» → «Торговое помещение»;

* «жилой дом», «апартаменты» → «Жилое здание»;

* всё остальное → «Иное сооружение».

```python
building = input("Введите тип здания: ").lower()

education_list = ["школа", "детский сад", "университет"]
shop_list = ["торговый центр", "магазин"]
appartment_list = ["жилой дом", "апартаменты"]

if building in education_list:
    category = "Образовательное учреждение"
elif building in shop_list:
    category = "Торговое помещение"
elif building in appartment_list:
    category = "Жилое здание"
else:
    category = "Иное сооружение"

print(f"Категория: {category}")
```

#### Задача 2. Подсчёт этажей с учётом ограничений

В городе действует ограничение:

* в исторической зоне — не выше 5 этажей;

* в новой застройке — не выше 20 этажей.

Программа:

1. Запрашивает зону («историческая» или «новая»).

2. Запрашивает количество этажей.

3. С помощью ветвления проверяет, соответствует ли этажность норме.

4. Выводит: «Допустимо» или «Превышение нормы!».

```python
zone = input("Зона (историческая/новая): ").lower()
floors = int(input("Количество этажей: "))


if zone == "историческая" and floors <= 5:
    print("Допустимо")
elif zone == "новая" and floors <= 20:
    print("Допустимо")
else:
    print("Превышение нормы!")
```

#### Задача 3. Распределение зелёных зон по районам

Есть 3 района города с разной площадью зелёных зон (в га):

* район A: 150 га;

* район B: 80 га;

* район C: 200 га.

Что делает программа?  Перебирает районы.

Для каждого района:

1. Если площадь ≥ 150 га → выводит «Высокий уровень озеленения»;

2. Если 50 ≤ площадь < 150 → «Средний уровень»;

3. Иначе → «Низкий уровень».

```python
areas = {"A": 150, "B": 80, "C": 200}


for district, green_area in areas.items():
    if green_area >= 150:
        level = "Высокий уровень озеленения"
    elif green_area >= 50:
        level = "Средний уровень"
    else:
        level = "Низкий уровень"
    print(f"Район {district}: {level}")
```

#### Задача 8. Поиск районов с высокой плотностью населения
Есть список районов с данными: площадь (км²) и численность населения (тыс. чел.). Нужно:
* в цикле перебрать все районы;
* для каждого вычислить плотность населения (население / площадь);
* если плотность превышает 5 тыс. чел./км², добавить район в список «плотно заселённых»;
* в конце вывести список таких районов.

```python
districts_data = [
    ("Центр", 10, 60),      # площадь 10 км², население 60 тыс. чел.
    ("Спальный", 25, 80),   # площадь 25 км², население 80 тыс. чел.
    ("Промзона", 15, 45),   # площадь 15 км², население 45 тыс. чел.
    ("Новый", 30, 75),     # площадь 30 км², население 75 тыс. чел.
    ("Старый", 8, 50)       # площадь 8 км², население 50 тыс. чел.
]

dense_districts = []  # список плотно заселённых районов

for name, area, population in districts_data:
    density = population / area  # плотность: тыс. чел. на км²
    if density > 5:
        dense_districts.append(name)

print("Плотно заселённые районы:", dense_districts)
```

#### Задача 9. Баланс зелёных зон и застройки

В городе должна соблюдаться норма: не менее 30 % площади — зелёные зоны. Программа:

1. Запрашивает общую площадь района (в га).

2. В цикле while запрашивает площадь очередной застройки (в га) до ввода 0.

3. На каждом шаге:

* добавляет площадь застройки к суммарной;

* вычисляет текущую долю зелёных зон: (общая_площадь − суммарная_застройка) / общая_площадь;

* если доля зелёных зон < 30 %, выводит предупреждение: *Нарушение нормы зелёных зон!* и прерывает цикл.

4. Если цикл завершился без нарушений, выводит: *Нормы соблюдены*.

```python
# Ввод общей площади района
total_area = float(input("Общая площадь района (га): "))


# Инициализация переменных
built_area = 0.0
violation_occurred = False  # Флаг: было ли нарушение нормы


# Цикл ввода площадей застройки (завершается только при вводе 0)
building_area = float(input("Площадь застройки (га): "))


while building_area != 0:
    built_area += building_area
    green_ratio = (total_area - built_area) / total_area * 100
    
    # Проверяем норму: если зелёных зон < 30%, ставим флаг
    if green_ratio < 30:
        violation_occurred = True
    
    # Следующий ввод
    building_area = float(input("Площадь застройки (га): "))


# Вывод результата
if violation_occurred:
    print("Нарушение нормы зелёных зон!")
else:
    print("Нормы соблюдены.")
```

#### Задача 10. Моделирование роста города с ограничениями

Замоделируйте рост населения города на 30 лет (с 2025 по 2055). Каждый год население увеличивается на 5 %.

Что нужно:

* Запросить у пользователя начальное население (в тыс. чел.).

* В цикле while вывести население за каждый год (округляя до целых тыс.).

* Если население достигнет 1 млн чел. раньше 2055 года — сообщить об этом.

* Если за 30 лет миллион не набран — вывести соответствующее сообщение.

```python
population = float(input("Начальное население (тыс. чел.): "))
year = 2025
end_year = 2055
reached_million = False


while year < end_year and not reached_million:
    population = int(population * 1.05)  # +5%, округление до целых
    print(f"Год {year}: население {population} тыс. чел.")
    
    if population >= 1000:
        reached_million = True
        print(f"Город достиг миллиона жителей в {year} году!")
    
    year += 1

if not reached_million:
    print("За 30 лет город не достиг миллиона жителей.")
```
