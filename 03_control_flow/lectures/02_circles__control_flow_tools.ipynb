#### Задача. Контроль плотности застройки

Программа моделирует застройку квартала. На каждом шаге (while) она:

* запрашивает площадь очередного здания (в м²);

* добавляет её к общей площади застройки;

* проверяет, не превысила ли суммарная площадь 10 000 м².

Когда площадь превышена, цикл прерывается и выводится сообщение:

*Превышение допустимой плотности застройки!*

В конце также выводится итоговая площадь.

#### Задача. Баланс зелёных зон и застройки

В городе должна соблюдаться норма: не менее 30 % площади — зелёные зоны. Программа:

1. Запрашивает общую площадь района (в га).

2. В цикле while запрашивает площадь очередной застройки (в га) до ввода 0.

3. На каждом шаге:

* добавляет площадь застройки к суммарной;

* вычисляет текущую долю зелёных зон: (общая_площадь − суммарная_застройка) / общая_площадь;

* если доля зелёных зон < 30 %, выводит предупреждение: *Нарушение нормы зелёных зон!* и прерывает цикл.

4. Если цикл завершился без нарушений, выводит: *Нормы соблюдены*.

#### Конструкции для обработки исключений (try, except, finally).

**Исключения (exceptions)** — это механизм обработки ошибок во время выполнения программы. Они позволяют программе продолжить работу после обнаружения ошибки, а не завершаться аварийно. В Python есть встроенные исключения, которые обрабатывают большинство типовых ошибок.

*TypeError* — операция или функция применяется к объекту несоответствующего типа

*ValueError* — операция или функция получает аргумент неподходящего значения. К примеру, исключение возникает, если попытаться преобразовать строку в число.

*IndexError* — обращение к элементу по несуществующему индексу

*ZeroDivisionError* — деление числа на ноль.

*FileNotFoundError* — Python не может найти файл, который мы хотим открыть.

```python
with open('non_existent_file.txt', 'r') as f:
    print(f.read())
```

#### Обработка исключений

С помощью блока *try* Python проверяет код на наличие исключений. Если в *try* встречается ошибка, выполнение переходит к первому блоку **except*.

```python
try:
    print(undefined_variable)
except NameError:
   print('Переменная не определена.')
```

В *finally* помещают код, который будет выполняться независимо от того, была ли найдена ошибка или нет. Часто этот блок используют для работы с файлами, чтобы закрыть документ.

*else* выполняется, если try не нашёл исключений.

Ключевое слово *as* при обработке ошибок используется для присвоения исключению переменной.

Команда *raise* в Python используется для принудительного вызова исключения. Это может быть полезно, если мы столкнулись с условием, которое должно остановить выполнение программы или вызвать ошибку.

Блок *except* без указания конкретного типа исключения будет обрабатывать все исключения, которые не были обработаны в предыдущих блоках except. 

Но так обычно не делают. Вместо этого разработчики предпочитают *except Exception*.

* Исключения в Python полезны для написания отказоустойчивого кода.
