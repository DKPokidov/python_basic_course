#### Задача 1. Генератор натуральных чисел
Напишите функцию‑генератор count_up_to(n), которая:
* принимает целое число n (предполагается, что n > 0);
* по очереди выдаёт числа от 1 до n (включительно);
* использует оператор yield.

Требования
* Не создавайте список всех чисел заранее — генерируйте каждое число «на лету».
* Используйте цикл внутри функции и yield для возврата каждого значения.
* Проверьте работу генератора с помощью цикла for.

```python
def count_up_to(n):
    count = 0
    while n != count:
        yield count
        count += 1

counter = count_up_to(8)
for num in counter:
    print(num)
```

#### Задача 2. Генератор приветствий

Напишите функцию‑генератор *greetings(names)*, которая:
* принимает список имён names;
* по очереди выдаёт строки вида "Привет, {имя}!" для каждого имени из списка;
* использует *yield* для возврата каждого приветствия.

Требования
* В цикле перебирайте имена из входного списка и для каждого формируйте строку с приветствием.
* Используйте *yield*, чтобы вернуть каждое приветствие и «заморозить» генератор до следующего вызова.
* Функция должна работать с любым непустым списком имён.

```python
def greetings(names):
    for name in names:
        yield f"Привет, {name}!"


# Проверка
names = ["Анна", "Борис", "Светлана"]
for greeting in greetings(names):
    print(greeting)
```

#### Задача 3. Написать итератор, который возвращает все простые числа в заданном диапазоне

```python
class PrimeIterator:
    def __init__(self, start, end):
        self.current = start
        self.end = end

    def __iter__(self):
        return self

    def __next__(self):
        # Ищем простое число, начиная с self.current
        while self.current < self.end:
            if self.is_prime(self.current):
                result = self.current
                self.current += 1
                return result  # Возвращаем найденное простое число
            self.current += 1  # Переходим к следующему числу
        raise StopIteration  # Если простые числа закончились

    def is_prime(self, num: int) -> bool:
        """Проверяет, является ли число num простым."""
        if num < 2:
            return False
        if num == 2:
            return True
        if num % 2 == 0:
            return False
        # Проверяем делители от 3 до √n (только нечётные)
        for i in range(3, int(num**0.5) + 1, 2):
            if num % i == 0:
                return False
        return True
    
pi = PrimeIterator(1, 50)

for i in pi:
    print(i)
```

#### Задача 4**. Генератор «зашумлённых» измерений
Условие

Напишите генераторную функцию *noisy_measurements(base_values, noise_level=0.1)*, которая:
* принимает список базовых значений *base_values* и уровень шума *noise_level* (по умолчанию 0.1);
* для каждого значения из *base_values* генерирует три «зашумлённых» измерения;
* каждое измерение вычисляется как:

##### **измерение = исходное_значение * (1 + случайное_отклонение)** ,
где случайное_отклонение — число из равномерного распределения на отрезке *[-noise_level, +noise_level]*;

* использует *yield* для выдачи значений по одному;
* не создаёт промежуточных списков (всё вычисляется «на лету»).

Требования:
* Используйте модуль *random* (функция random.uniform(a, b)).
* Генератор должен выдавать значения в порядке:

    **base_values[0]_изм1, base_values[0]_изм2, base_values[0]_изм3,
    base_values[1]_изм1, … и т. д.**

* При пустом *base_values* генератор не выдаёт ничего.

```python
import random

def noisy_measurements(base_values, noise_level=0.1):
    for value in base_values:
        for _ in range(3):  # Три измерения на каждое базовое значение
            # Генерируем случайное отклонение в диапазоне [-noise_level, +noise_level]
            deviation = random.uniform(-noise_level, noise_level)
            # Вычисляем зашумлённое значение
            noisy_value = value * (1 + deviation)
            yield noisy_value  # Выдаём по одному
```

```python
random.seed(42)  # Для воспроизводимости

list(noisy_measurements([10, 20], 0.05))
# Примерный вывод (зависит от seed):
# [10.03, 9.97, 10.01, 20.08, 19.92, 20.04]
```
