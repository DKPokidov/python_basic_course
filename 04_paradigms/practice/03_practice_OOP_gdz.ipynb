#### Задача 1. Разминка: моделируем городской район
Создайте класс CityDistrict для моделирования городского района. У района есть:

* название (name);

* площадь в гектарах (area_ha);

* население (population);

* список типов застройки (жилая, коммерческая, промышленная, зелёная зона) — хранится как список строк (building_types).

Требования к классу:

1. Конструктор (__init__): принимает name, area_ha, population и опционально building_types (по умолчанию — пустой список).

2. Метод add_building_type(type_name): добавляет новый тип застройки в список (если его ещё нет).

3. Метод get_density(): возвращает плотность населения (чел./га) как population / area_ha.

4. Метод has_green_space(): возвращает True, если в building_types есть 'green', иначе False.

5. Метод __str__: возвращает строку вида:

*Район: [name], площадь: [area_ha] га, население: [population] чел.*


```python
class CityDistrict():

    def __init__(self, name, area_ha, population, building_types=None):
        self.name = name
        self.area_ha = area_ha
        self.population = population
        self.building_types = building_types if building_types != None else []

    def __str__ (self):
        return f'Район: {self.name}, площадь: {self.area_ha} га, население: {self.population} чел.'

    def add_building_type(self, type_name):
        if type_name not in self.building_types:
            self.building_types.append(type_name)
            print("The new building type's added to building_types")
        else:
            print('building_types contains the new building type')

    def building_types_show(self):
        print(self.building_types)

    def get_density(self):
        return self.population / self.area_ha
    
    def has_green_space(self):
        green_str = 'green'
        return green_str in self.building_types
    
centre = CityDistrict(
    name='Центральный',
    area_ha=450,
    population=350e3,
    building_types='деловой'
)

str(centre)
```

#### Задача 3. Продолжение: район города
Создайте класс District, который содержит список зданий (buildings). Добавьте методы:
* add_building(building) — добавляет здание в район;
* get_total_height() — считает суммарную высоту всех зданий в районе;
* get_buildings_by_type(building_type) — возвращает список зданий определённого типа.

Используйте класс *Building* из **Задачи 2**.

```python
class District:
    def __init__(self) -> list:
        self.buildings = []

    def add_building(self, building: object):
        self.buildings.append(building)
    
    def get_total_height(self) -> int: 
        return sum(building.height for building in self.buildings)

    def get_buildings_by_type(self, building_type):
        building_type_list = []
        for building in self.buildings:
            if building_type == building.type:
                building_type_list.append(building.name)
        
        return building_type_list
```

```python
ofice = Building('ТЦ Адмирал', 30, 1998, 'офисный')
hospital = Building('36 больницы', 15, 1954, 'государственный')
house = Building('ЖК Жемчужина', 50, 2010, 'жилой')
school = Building('13 школа', 12, 1974, 'государственный')

centre = District()

centre.add_building(ofice)
centre.add_building(hospital)
centre.add_building(house)
centre.add_building(school)

height = centre.get_total_height()
print(height)

living = centre.get_buildings_by_type('государственный')
print(living)
```

#### Задача 4. Типы зданий
Условие: создайте подклассы для *Building*: *ResidentialBuilding* (жилое здание), *OfficeBuilding* (офисное здание), *ShoppingCenter* (торговый центр). Каждый подкласс должен иметь дополнительный атрибут:
* для *ResidentialBuilding*: *number_of_apartments* (количество квартир);
* для *OfficeBuilding*: *number_of_floors* (количество этажей);
* для *ShoppingCenter*: *number_of_shops* (количество магазинов).

Переопределите метод *get_info()* для каждого подкласса, чтобы он выводил дополнительную информацию.

```python
class ResidentialBuilding(Building):
    def __init__(self, name, height, year_built, number_of_apartments):
        super().__init__(name, height, year_built, "жилой")
        self.number_of_apartments = number_of_apartments

    def get_info(self):
        return f"{super().get_info()}, квартир: {self.number_of_apartments}"

class OfficeBuilding(Building):
    def __init__(self, name, height, year_built, number_of_floors):
        super().__init__(name, height, year_built, "офисный")
        self.number_of_floors = number_of_floors

    def get_info(self):
        return f"{super().get_info()}, этажей: {self.number_of_floors}"

class ShoppingCenter(Building):
    def __init__(self, name, height, year_built, number_of_shops):
        super().__init__(name, height, year_built, "торговый")
        self.number_of_shops = number_of_shops

    def get_info(self):
        return f"{super().get_info()}, магазинов: {self.number_of_shops}"
```

#### Задача 5*. Университет и студенты

Cоздайте классы *Student* (Студент) и *University* (Университет). 

Student должен иметь атрибуты: *name, group, average_grade* (средний балл). Добавьте метод *improve_grade(points)*, который увеличивает средний балл на *points* (с ограничением до 5.0).  

University должен содержать список студентов и методы:
* *enroll_student(student)* — ;
* *get_top_students(n)* — возвращает n студентов с наивысшим средним баллом;
* *get_students_by_group(group_name)* — возвращает студентов определённой группы.

```python
class Student:
    def __init__(self, name: str, group: str, average_grade: float):
        self.name = name
        self.group = str(group)
        self.average_grade = average_grade

    def improve_grade(self, points: float) -> float:
        self.average_grade = self.average_grade + points if self.average_grade + points < 5 else 5
        return self.average_grade

class University:
    def __init__(self):
        self.student_list = []
            
    def enroll_student(self, student):
        '''зачисляет студента'''
        self.student_list.append(student)
        print(f'Студент {student.name} зачислен группу {student.group}!')


    def get_top_students(self, n):
        '''возвращает n студентов с наивысшим средним баллом'''
        sorted_list = sorted(self.student_list, key=lambda student: student.average_grade, reverse=True) 
        print(f'Топ-{n} лучших студентов:')
        index = 0
        for student in sorted_list[:n]:
            index += 1
            print(f'{index}. {student.name}, средний балл: {student.average_grade}')

    def get_students_by_group(self, group_name):
        '''возвращает студентов определённой группы'''
        filtred_list = filter(lambda student: student.group == str(group_name), self.student_list)
        index = 0
        print(f'В группе {group_name} учатся:')
        for student in filtred_list:
            index += 1
            print(f'{index}. {student.name}')
```

```python
class Student:
    def __init__(self, name, group, average_grade):
        self.name = name
        self.group = group
        self.average_grade = min(average_grade, 5.0)  # Ограничение до 5.0

    def improve_grade(self, points):
        self.average_grade = min(self.average_grade + points, 5.0)

    def __repr__(self):
        return f"Студент {self.name}, группа {self.group}, средний балл {self.average_grade}"

class University:
    def __init__(self, name):
        self.name = name
        self.students = []

    def enroll_student(self, student):
        self.students.append(student)

    def get_top_students(self, n):
        sorted_students = sorted(self.students, key=lambda x: x.average_grade, reverse=True)
        return sorted_students[:n]

    def get_students_by_group(self, group_name):
        return [student for student in self.students if student.group == group_name]

# Пример использования
uni = University("МегаУниверситет")
student_1 = Student("Иван", "МФ-1", 4.5)
student_2 = Student("Мария", "МФ-1", 4.8)
uni.enroll_student(student_1)
uni.enroll_student(student_2)
print(uni.get_top_students(1))  # [Студент Мария, группа МФ-1, средний балл 4.8]
```

Задача 6*. Транспортные средства

Создайте базовую систему транспортных средств, используя наследование. Ваша цель — реализовать общую логику в родительском классе и специализировать её в дочерних.

Требования:

1. Создайте родительский класс *Vehicle* со следующими атрибутами и методами:

* атрибуты: *brand* (марка), *model* (модель), *year* (год выпуска);

* метод *start_engine()* — выводит сообщение «Двигатель запущен»;

* метод *info()* — возвращает строку с базовой информацией о транспортном средстве.

2. Создайте два дочерних класса, наследующих *Vehicle*:

*Car* (автомобиль):

* добавляет атрибут *doors* (количество дверей);

* переопределяет *info()*, чтобы включать информацию о количестве дверей.

*Motorcycle* (мотоцикл):

* добавляет атрибут *has_sidecar* (есть ли коляска, *True/False*);

* переопределяет *info()*, чтобы указывать наличие коляски.

3. В обоих дочерних классах переопределите метод *start_engine()*:

* для *Car*: выводит «Автомобиль [марка] [модель] завёлся с характерным звуком»;

* для *Motorcycle*: выводит «Мотоцикл [марка] [модель] рычит при запуске».

4. Создайте по одному объекту каждого класса, вызовите для них методы *start_engine()* и *info()*.

```python
class Vehicle:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year

    def start_engine(self):
        print("Двигатель запущен")

    def info(self):
        return f"{self.year} {self.brand} {self.model}"

class Car(Vehicle):
    def __init__(self, brand, model, year, doors):
        super().__init__(brand, model, year)
        self.doors = doors

    def start_engine(self):
        print(f"Автомобиль {self.brand} {self.model} завёлся с характерным звуком")

    def info(self):
        return f"{super().info()}, дверей: {self.doors}"

class Motorcycle(Vehicle):
    def __init__(self, brand, model, year, has_sidecar):
        super().__init__(brand, model, year)
        self.has_sidecar = has_sidecar

    def start_engine(self):
        print(f"Мотоцикл {self.brand} {self.model} рычит при запуске")

    def info(self):
        sidecar_info = "с коляской" if self.has_sidecar else "без коляски"
        return f"{super().info()}, {sidecar_info}"

# Пример использования
car = Car("Toyota", "Camry", 2023, 4)
motorcycle = Motorcycle("Harley-Davidson", "Street 750", 2022, True)

car.start_engine()
print(car.info())

motorcycle.start_engine()
print(motorcycle.info())
```
