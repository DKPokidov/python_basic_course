```python
import json

with open('../../data/city_data.txt', 'r', encoding='utf-8') as f:
    city_data = json.load(f)
```

```python
city_data
```

```python
# 1. Фильтрация по типу объекта

def filtre_by_type(data: dict, type: str) -> list:
    '''Фильтрация по типу объекта'''
    return [data['name'] for data in data['objects'] 
            if data['type'] == type] 
```

```python
obj_type = 'kindergarten'
filtre_by_type(data=city_data, type=obj_type)
```

```python
# 2. Анализ доступности

def accessibility_analysis(data: dict) -> dict:
    '''Анализ доступности'''
    result = {}  
    for obj in data['objects']:
        if obj['accessibility']['ramp'] is True:
            district = obj['district']
            result[district] = result.get(district, 0) + 1
        
    return result
```

```python
accessibility_analysis(data = city_data)
```

```python
# 3. Поиск перегруженных объектовЫ

def find_overloaded(data: dict, threshold: int) -> list:
    '''Поиск перегруженных объектов'''
    for obj in data['objects']:
        if obj['capacity'] > threshold:
            print(f"{obj['name']} ({obj['capacity']} мест)")
```

```python
# нужно получить name и capacity
find_overloaded(data=city_data, threshold=500)
```

```python
# 4. Статистика по сервисам

def services_stats(data: dict) -> dict:
    '''Статистика по сервисам'''
    obj = data['objects']
    object_types = {dictionary['type']  for dictionary in obj}
    final_dict = {}

    for type in object_types:
        type_set = set()
        type_dict = list(filter(lambda obj: obj['type'] == type, obj))
        for service in type_dict:
            type_set.update(service['services'])
        final_dict[type] = list(type_set)

    return final_dict
```

```python
services_stats(data=city_data)
```
